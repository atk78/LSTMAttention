```python
import pandas as pd
from rdkit import Chem

permitted_atomes = ["B", "C", "N", "O", "S", "P", "F", "Cl", "Br", "I"] + ["*"]
permitted_atomes_set = set(permitted_atomes)

molecule_of_polymer = "polymer"
csv_file_name = "Khazana_dc"
property_name = "dielectric_constant"
file_path = f"./../../../data/raw/{molecule_of_polymer}/{csv_file_name}.csv"

smiles_col_name = "smiles"
property_col_name = "dielectric_constant"

df = pd.read_csv(file_path)
df = df[["smiles", "value"]].rename(
    columns={"smiles": smiles_col_name, "value": property_col_name})
print("前処理前のサンプル数:", len(df))
def check_atoms(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)
        atoms_set = set([atom.GetSymbol() for atom in mol.GetAtoms()])
        isin = permitted_atomes_set > atoms_set
        if isin:
            return smiles
        else:
            return None
    except:
        return None

df[smiles_col_name] = df[smiles_col_name].map(check_atoms)
df = df.dropna().reset_index(drop=True)
df.to_csv(f"./../../../data/processed/{molecule_of_polymer}/{property_name}.csv", index=False)
print("前処理後のサンプル数:", len(df))
```
